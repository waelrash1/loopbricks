# Glossary {#sec-glossary}

The terms the book uses, in plain words, with the chapter or appendix that explains each one. Related entries are listed under *See also*. Statistics and machine learning terms are here too, so that a reader coming from either side can look up the other.

Abbreviations: MAE, mean absolute error; MAPE, mean absolute percentage error; RMSE, root mean square error.

## A

[**Acquisition delay**]{#gloss-acquisition-delay}
:   The average time between ordering something and receiving it. Orders in that time sit in the supply line. Explained in @sec-swings. *See also* [Supply line](#gloss-supply-line).

[**Adjustment time**]{#gloss-adjustment-time}
:   How long a goal-seeking rule takes to close a gap: each period it closes the gap divided by the adjustment time. A short adjustment time is a hard push; a long one a gentle push. Explained in @sec-loops. *See also* [Goal seeking](#gloss-goal-seeking), [Time constant](#gloss-time-constant).

[**Ageing chain**]{#gloss-ageing-chain}
:   A row of stocks that things pass through in order, such as age bands or waiting-time bands, each stage feeding the next. It tracks how a stock is spread across stages, not just its total. Explained in @sec-delays. *See also* [Order of a delay](#gloss-order-of-a-delay).

[**Agent**]{#gloss-agent}
:   A language model that works on a task in a loop: it chooses a tool (a Python function), reads the result and decides the next step, until it gives an answer. The tools compute every number. Explained in @sec-ai-partners. *See also* [Tool](#gloss-tool), [Large language model](#gloss-large-language-model).

[**Anchoring and adjustment**]{#gloss-anchoring-and-adjustment}
:   A judgement made by starting from a reference value and adjusting it for evidence. When the anchor itself drifts towards recent experience, goals erode. Explained in @sec-decisions. *See also* [Eroding goals](#gloss-eroding-goals).

[**Archetype**]{#gloss-archetype}
:   A small loop structure that recurs across fields and produces a recognisable behaviour, a typical trap and a typical way out, such as fixes that fail or limits to growth. An archetype suggests a hypothesis to test, not a conclusion. Explained in @sec-patterns. *See also* [Leverage point](#gloss-leverage-point).

[**Autocorrelation**]{#gloss-autocorrelation}
:   The correlation of a series with itself some number of steps earlier (the lag). Residuals that are autocorrelated mean a model is missing something, and that methods assuming independent errors will be too confident. Explained in @sec-learning. *See also* [Correlation coefficient](#gloss-correlation-coefficient), [Residuals](#gloss-residuals).

[**Automation bias**]{#gloss-automation-bias}
:   Accepting an automated answer, such as the first policy an agent proposes, without checking it. The remedy is a check that runs whether or not the answer looks right. Explained in @sec-ai-decide. *See also* [Red team](#gloss-red-team).

[**Autoregression**]{#gloss-autoregression}
:   A forecasting model that predicts the next value of a series from its own recent values, fitted by least squares. Explained in @sec-learning. *See also* [Machine learning](#gloss-machine-learning).

[**Auxiliary variable**]{#gloss-auxiliary-variable}
:   A variable worked out at each moment from stocks, constants and other variables, with no memory of its own. In the `loopbricks` spec these are the `vars`. Explained in @sec-push-back. *See also* [Constant](#gloss-constant), [Stock](#gloss-stock).

## B

[**Balancing loop**]{#gloss-balancing-loop}
:   A feedback loop that counteracts change, pulling a stock towards a goal or limit. It has an odd number of negative links and is labelled B. Explained in @sec-push-back. *See also* [Goal seeking](#gloss-goal-seeking), [Reinforcing loop](#gloss-reinforcing-loop).

[**Bass diffusion model**]{#gloss-bass-diffusion-model}
:   A model of how a new product or idea spreads: adoption driven by outside influence (innovation) and by contact with adopters (imitation), giving an S-shaped curve. Explained in @sec-loops. *See also* [Logistic growth](#gloss-logistic-growth).

[**Benchmark forecast**]{#gloss-benchmark-forecast}
:   A deliberately simple forecast, such as 'the same as last month' or 'the historical average', that any serious model must beat on data it has not seen. Explained in @sec-learning. *See also* [Out-of-sample test](#gloss-out-of-sample-test).

[**Bootstrap**]{#gloss-bootstrap}
:   A way to estimate uncertainty by resampling the data with replacement many times, recomputing the estimate each time and reading the spread. The block bootstrap resamples runs of consecutive points, to keep their autocorrelation. Explained in @sec-learning. *See also* [Confidence interval](#gloss-confidence-interval).

[**Boundary**]{#gloss-boundary}
:   The line between what a model explains (endogenous), what it takes as given (exogenous) and what it leaves out. A model boundary chart lists all three. Explained in @sec-story-to-diagram. *See also* [Endogenous](#gloss-endogenous), [Exogenous](#gloss-exogenous).

[**Bounded rationality**]{#gloss-bounded-rationality}
:   Decision making with limited information, time and attention, so that people use rules of thumb rather than optimise. Decision rules in system dynamics models aim to describe these rules. Explained in @sec-decisions. *See also* [Decision rule](#gloss-decision-rule).

## C

[**Calibration**]{#gloss-calibration}
:   Choosing a model's uncertain parameters so that its output matches observed data as closely as possible, usually by minimising a payoff function. Explained in @sec-calibration. *See also* [Payoff function](#gloss-payoff-function), [Identifiability](#gloss-identifiability).

[**Carrying capacity**]{#gloss-carrying-capacity}
:   The level at which a limited resource stops a growing stock: in logistic growth, the level where net growth falls to zero. Explained in @sec-loops. *See also* [Logistic growth](#gloss-logistic-growth).

[**Causal link**]{#gloss-causal-link}
:   An arrow from a cause to an effect, marked + if a rise in the cause makes the effect higher than it would otherwise have been, and - if it makes it lower. Explained in @sec-story-to-diagram. *See also* [Loop polarity](#gloss-loop-polarity).

[**Causal loop diagram**]{#gloss-causal-loop-diagram}
:   A diagram of variables joined by signed causal links, showing the feedback loops that a story describes. It shows loops clearly but hides which variables are stocks. Explained in @sec-story-to-diagram. *See also* [Stock-and-flow diagram](#gloss-stock-and-flow-diagram).

[**Coflow**]{#gloss-coflow}
:   A second stock that moves alongside a main stock to track an attribute of what it holds, such as the total experience of a workforce, so that the average attribute can be followed over time. Explained in @sec-delays.

[**Confidence interval**]{#gloss-confidence-interval}
:   A range built from data so that, if the whole exercise were repeated many times, a stated share of such ranges (often 95%) would contain the true value. It describes uncertainty in an estimate, not the spread of future values. Explained in @sec-learning. *See also* [Prediction interval](#gloss-prediction-interval), [Standard error](#gloss-standard-error).

[**Conservation check**]{#gloss-conservation-check}
:   A test that everything that flowed into a stock either flowed out or is still there: the change in the stock equals the sum of net flow times the time step. Explained in @sec-push-back. *See also* [Verification](#gloss-verification).

[**Constant**]{#gloss-constant}
:   A model input that does not change during a run, such as a capacity or a delay time. In `loopbricks` the constants are the model's `params`, which `run()` can change. Explained in @sec-push-back. *See also* [Auxiliary variable](#gloss-auxiliary-variable), [Preset](#gloss-preset).

[**Correlation coefficient**]{#gloss-correlation-coefficient}
:   A number from -1 to 1 measuring how closely two series move together along a straight line. In a feedback system it is a poor guide to cause. Explained in @sec-learning. *See also* [Autocorrelation](#gloss-autocorrelation).

## D

[**Decision rule**]{#gloss-decision-rule}
:   The equation behind a controlled flow: how the actors in the system set it from the information they actually have, such as an ordering or hiring rule. Explained in @sec-decisions. *See also* [Bounded rationality](#gloss-bounded-rationality).

[**Delay**]{#gloss-delay}
:   A gap in time between a cause and its effect. A material delay holds things in transit; an information delay is a lag in what people believe. Explained in @sec-delays. *See also* [Material delay](#gloss-material-delay), [Information delay](#gloss-information-delay).

[**Dimensional consistency**]{#gloss-dimensional-consistency}
:   The requirement that every equation balances in units: flows are in stock units per unit of time, and both sides of each equation have the same units. Explained in @sec-testing. *See also* [Verification](#gloss-verification).

[**Doubling time**]{#gloss-doubling-time}
:   The time an exponentially growing quantity takes to double: about 70 divided by the growth rate in percent (the rule of 70). Explained in @sec-loops. *See also* [Exponential growth](#gloss-exponential-growth).

[**dt**]{#gloss-dt}
:   The time step of a simulation: how far the clock moves on each step. A good dt is small relative to the model's shortest time constant; halving it should barely change the results. Explained in @sec-push-back. *See also* [Euler's method](#gloss-euler-s-method), [Halving test](#gloss-halving-test).

[**Dynamic hypothesis**]{#gloss-dynamic-hypothesis}
:   A proposed explanation of the reference mode in terms of feedback structure: which loops and delays produce the behaviour. It is tested against data and against rival hypotheses. Explained in @sec-story-to-diagram. *See also* [Reference mode](#gloss-reference-mode).

## E

[**Effect**]{#gloss-effect}
:   A multiplier, usually 1 under normal conditions, that changes a quantity in response to a pressure such as workload or fatigue. Effects are often drawn as lookup tables. Explained in @sec-decisions. *See also* [Lookup](#gloss-lookup).

[**Endogenous**]{#gloss-endogenous}
:   Explained inside the model, by its own feedback structure. System dynamics aims to make the behaviour of interest endogenous. Explained in @sec-story-to-diagram. *See also* [Exogenous](#gloss-exogenous), [Boundary](#gloss-boundary).

[**Equilibrium**]{#gloss-equilibrium}
:   A state where every stock stays constant because its inflows equal its outflows. An equilibrium can be stable (disturbances die out) or unstable (they grow). Explained in @sec-loops. *See also* [Tipping point](#gloss-tipping-point).

[**Eroding goals**]{#gloss-eroding-goals}
:   An archetype in which a gap between goal and performance is closed by lowering the goal rather than improving performance. Explained in @sec-patterns. *See also* [Archetype](#gloss-archetype), [Anchoring and adjustment](#gloss-anchoring-and-adjustment).

[**Euler's method**]{#gloss-euler-s-method}
:   The simplest way to simulate stocks: at each step, work out the flows from the current stocks, then move each stock by its net flow times dt. Explained in @sec-push-back. *See also* [dt](#gloss-dt), [Integration error](#gloss-integration-error).

[**Exogenous**]{#gloss-exogenous}
:   Taken as given from outside the model, such as a demand series or a policy date. Explained in @sec-story-to-diagram. *See also* [Endogenous](#gloss-endogenous).

[**Exponential decay**]{#gloss-exponential-decay}
:   Decline in which the outflow is a fixed fraction of what remains, so the stock falls fast at first and ever more slowly, losing about 63% in one time constant. Explained in @sec-loops. *See also* [Time constant](#gloss-time-constant), [Half-life](#gloss-half-life).

[**Exponential growth**]{#gloss-exponential-growth}
:   Growth in which the increase is a fixed fraction of the current level, produced by a dominant reinforcing loop. The stock doubles in equal intervals. Explained in @sec-loops. *See also* [Doubling time](#gloss-doubling-time), [Reinforcing loop](#gloss-reinforcing-loop).

[**Extreme conditions test**]{#gloss-extreme-conditions-test}
:   Running a model with inputs pushed to extremes, such as zero capacity or a huge inflow, to check that it still behaves physically. Explained in @sec-push-back. *See also* [Validation](#gloss-validation).

## F

[**Feedback loop**]{#gloss-feedback-loop}
:   A closed chain of causes in which a change comes back to affect itself. Loops are reinforcing or balancing. Explained in @sec-push-back. *See also* [Balancing loop](#gloss-balancing-loop), [Reinforcing loop](#gloss-reinforcing-loop).

[**Fixes that fail**]{#gloss-fixes-that-fail}
:   An archetype in which a quick fix relieves a problem but sets off a delayed side effect that brings the problem back, often worse. Explained in @sec-patterns. *See also* [Archetype](#gloss-archetype), [Shifting the burden](#gloss-shifting-the-burden).

[**Floor at zero**]{#gloss-floor-at-zero}
:   A guard that stops a stock that cannot be negative from going below zero. A floor that is used means the flows and stock no longer add up, so it is better to limit the outflow in the equations. Explained in @sec-push-back. *See also* [Conservation check](#gloss-conservation-check).

[**Flow**]{#gloss-flow}
:   A rate of change of a stock, such as referrals or treatments per month. Inflows add to a stock and outflows drain it; drawn as a pipe with a valve. Explained in @sec-push-back. *See also* [Stock](#gloss-stock), [Net flow](#gloss-net-flow).

[**Fractional rate**]{#gloss-fractional-rate}
:   A flow expressed as a fraction of a stock per unit of time, such as a growth rate of 3% a year or one tenth of the list each month. Explained in @sec-loops. *See also* [Exponential growth](#gloss-exponential-growth).

## G

[**Generative AI**]{#gloss-generative-ai}
:   Models that produce text, code or images from instructions. In modelling they draft variable lists, diagrams, equations and briefs, which code then checks. Explained in @sec-ai-partners. *See also* [Large language model](#gloss-large-language-model), [Hallucination](#gloss-hallucination).

[**Goal seeking**]{#gloss-goal-seeking}
:   Behaviour in which a stock approaches a target, each step closing part of the remaining gap, produced by a balancing loop. Explained in @sec-loops. *See also* [Adjustment time](#gloss-adjustment-time), [Balancing loop](#gloss-balancing-loop).

[**Group model building**]{#gloss-group-model-building}
:   Building a model with the people who know the system, so that their knowledge shapes the structure and they own the conclusions. Explained in @sec-story-to-diagram.

## H

[**Half-life**]{#gloss-half-life}
:   The time a decaying stock takes to halve: about 0.69 times its time constant. Explained in @sec-loops. *See also* [Exponential decay](#gloss-exponential-decay).

[**Hallucination**]{#gloss-hallucination}
:   A fluent, confident output that is not true: an invented quote, reference, function or number. Every AI output in the book meets a check by code for this reason. Explained in @sec-ai-partners. *See also* [Unsourced number](#gloss-unsourced-number), [Generative AI](#gloss-generative-ai).

[**Halving test**]{#gloss-halving-test}
:   Running a model again with dt halved and checking that the results barely change; if they do, dt is too large. Explained in @sec-push-back. *See also* [dt](#gloss-dt).

[**Hybrid model**]{#gloss-hybrid-model}
:   A model that combines a feedback structure with a component learned from data, such as an effect estimated by machine learning. Explained in @sec-ai-test. *See also* [Machine learning](#gloss-machine-learning).

## I

[**Identifiability**]{#gloss-identifiability}
:   Whether the data can pin down a parameter. When two parameters can trade off against each other with the same fit, neither is identified, however good the fit. Explained in @sec-calibration. *See also* [Calibration](#gloss-calibration).

[**Information delay**]{#gloss-information-delay}
:   A lag in perception or belief: people act on a smoothed or out-of-date picture of the system. Unlike a material delay, nothing is conserved in it. Explained in @sec-delays. *See also* [Smoothing](#gloss-smoothing), [Material delay](#gloss-material-delay).

[**Integration error**]{#gloss-integration-error}
:   The difference between the simulated path and the exact one, caused by taking steps of finite size. With Euler's method it shrinks in proportion to dt. Explained in @sec-push-back. *See also* [Euler's method](#gloss-euler-s-method), [dt](#gloss-dt).

## L

[**Large language model**]{#gloss-large-language-model}
:   A program trained on a very large body of text to continue text. It follows instructions in plain language, but it predicts what is likely, not what is checked. Abbreviated LLM. Explained in @sec-ai-partners. *See also* [Generative AI](#gloss-generative-ai), [Reasoning model](#gloss-reasoning-model).

[**Leakage**]{#gloss-leakage}
:   Any route by which information from the test data reaches a model during training, making its test error look better than it will be in use. Explained in @sec-learning. *See also* [Training and test sets](#gloss-training-and-test-sets).

[**Least squares**]{#gloss-least-squares}
:   Fitting a model by choosing parameters that minimise the sum of squared residuals. Explained in @sec-learning. *See also* [Residuals](#gloss-residuals), [Calibration](#gloss-calibration).

[**Leverage point**]{#gloss-leverage-point}
:   A place in a system where a small change makes a large and lasting difference, often in a decision rule, a delay or a goal rather than in a number. Explained in @sec-patterns. *See also* [Archetype](#gloss-archetype).

[**Limits to growth**]{#gloss-limits-to-growth}
:   An archetype in which a reinforcing growth loop meets a balancing loop from a limit, so growth slows and stops. Explained in @sec-patterns. *See also* [Logistic growth](#gloss-logistic-growth).

[**Little's law**]{#gloss-little-s-law}
:   In a steady state, the average number waiting equals the arrival rate times the average time each one waits. Explained in @sec-delays. *See also* [Residence time](#gloss-residence-time).

[**Logistic growth**]{#gloss-logistic-growth}
:   S-shaped growth in which a reinforcing loop dominates at first and a balancing loop takes over as the stock nears its carrying capacity. Net growth is fastest at half the capacity. Explained in @sec-loops. *See also* [Carrying capacity](#gloss-carrying-capacity), [Shifting dominance](#gloss-shifting-dominance).

[**Lookup**]{#gloss-lookup}
:   A function of one input drawn as points joined by straight lines and held flat beyond the ends; `TABLE` in the equation language. Used for effects that are judged rather than derived. Explained in @sec-decisions. *See also* [Effect](#gloss-effect).

[**Loop dominance**]{#gloss-loop-dominance}
:   Which loop drives a stock's behaviour at a given moment. Behaviour changes character when dominance shifts from one loop to another. Explained in @sec-loops. *See also* [Shifting dominance](#gloss-shifting-dominance).

[**Loop polarity**]{#gloss-loop-polarity}
:   Whether a loop reinforces or balances: count its negative links; an even number gives a reinforcing loop, an odd number a balancing one. Explained in @sec-story-to-diagram. *See also* [Causal link](#gloss-causal-link).

## M

[**Machine learning**]{#gloss-machine-learning}
:   Fitting flexible models to data so that they learn a relationship from examples, judged by their error on data they have not seen. Explained in @sec-learning. *See also* [Overfitting](#gloss-overfitting), [Training and test sets](#gloss-training-and-test-sets).

[**Magic constant**]{#gloss-magic-constant}
:   A number written inside an equation with no name, units or source, such as 1.2 in a capacity rule. It hides an assumption; give it a name and a source. Explained in @sec-ai-formulate. *See also* [Decision rule](#gloss-decision-rule).

[**Material delay**]{#gloss-material-delay}
:   A delay in which things are conserved while in transit, such as orders in a pipeline or patients in treatment. Every material delay hides a stock. Explained in @sec-delays. *See also* [Delay](#gloss-delay), [Order of a delay](#gloss-order-of-a-delay).

[**Mean absolute error**]{#gloss-mean-absolute-error}
:   The average size of a model's misses, in the data's own units (MAE). Explained in @sec-learning. *See also* [Root mean square error](#gloss-root-mean-square-error).

[**Mean absolute percentage error**]{#gloss-mean-absolute-percentage-error}
:   The average size of a model's misses as a share of the data (MAPE). Easy to explain, but misleading when the data come near zero. Explained in @sec-learning. *See also* [Mean absolute error](#gloss-mean-absolute-error).

[**Molecule**]{#gloss-molecule}
:   A small, reusable model of one recurring structure, such as a stock that drains in proportion to itself. The book's molecules are our own implementations of structures described by Hines; @sec-molecules lists them all. Explained in @sec-push-back.

[**Monte Carlo analysis**]{#gloss-monte-carlo-analysis}
:   Running a model many times with inputs drawn at random from their plausible ranges, to see the spread of outcomes they produce. Explained in @sec-testing. *See also* [Sensitivity analysis](#gloss-sensitivity-analysis).

[**Moving target**]{#gloss-moving-target}
:   A goal that depends on the very stock the decision changes, so that pursuing it shifts it. Explained in @sec-swings.

[**Mutation testing**]{#gloss-mutation-testing}
:   Planting a known fault in a model to measure whether a test plan would catch it. A plan that catches no planted faults tests only what passes. Explained in @sec-ai-test. *See also* [Verification](#gloss-verification).

## N

[**Net flow**]{#gloss-net-flow}
:   Inflows minus outflows: the rate at which a stock changes. A stock series reveals only the net flow, never the separate flows. Explained in @sec-push-back. *See also* [Flow](#gloss-flow).

[**Normal distribution**]{#gloss-normal-distribution}
:   The bell-shaped distribution that arises when many small, independent effects add up. About 68% of values lie within one standard deviation of the mean and 95% within two. Explained in @sec-learning. *See also* [Standard deviation](#gloss-standard-deviation).

## O

[**Order of a delay**]{#gloss-order-of-a-delay}
:   The number of stages a delay has. A first-order delay releases things at once and tails off; higher orders release them closer to the average delay time. The order sets the shape of the response, not its average. Explained in @sec-delays. *See also* [Ageing chain](#gloss-ageing-chain), [Material delay](#gloss-material-delay).

[**Oscillation**]{#gloss-oscillation}
:   Repeated swings above and below a goal, typically produced by a balancing loop acting through a delay it does not account for. Explained in @sec-swings. *See also* [Overshoot](#gloss-overshoot), [Period](#gloss-period).

[**Out-of-sample test**]{#gloss-out-of-sample-test}
:   Measuring a model's error on data it was not fitted to, such as later years held back. For time series the split must be in time. Explained in @sec-learning. *See also* [Training and test sets](#gloss-training-and-test-sets).

[**Overfitting**]{#gloss-overfitting}
:   Fitting a model so flexibly that it learns the noise in the training data as well as the signal, so it fits the past well and forecasts badly. Explained in @sec-learning. *See also* [Out-of-sample test](#gloss-out-of-sample-test).

[**Overshoot**]{#gloss-overshoot}
:   Passing a goal or limit before turning back, because the response to a gap arrives after a delay. Explained in @sec-swings. *See also* [Oscillation](#gloss-oscillation).

## P

[**Path dependence**]{#gloss-path-dependence}
:   When small early differences, amplified by reinforcing loops, decide where a system ends up, so history matters. Explained in @sec-loops. *See also* [Tipping point](#gloss-tipping-point).

[**Payoff function**]{#gloss-payoff-function}
:   The measure of fit that calibration minimises, usually a weighted sum of squared differences between model output and data. Explained in @sec-calibration. *See also* [Calibration](#gloss-calibration), [Least squares](#gloss-least-squares).

[**Percentile**]{#gloss-percentile}
:   The value below which a given share of the data falls; the 5th and 95th percentiles bracket the middle 90%. Explained in @sec-learning.

[**Period**]{#gloss-period}
:   The time between successive peaks of an oscillation. In system dynamics models it is set largely by the delays in the loop. Explained in @sec-swings. *See also* [Oscillation](#gloss-oscillation).

[**Policy**]{#gloss-policy}
:   A change to a decision rule or a controllable constant, tested in the model before it is tried in the world. Explained in @sec-model-to-decision. *See also* [Robust policy](#gloss-robust-policy).

[**Policy resistance**]{#gloss-policy-resistance}
:   The tendency of a system to counteract an intervention, because feedback loops respond to the change. Explained in @sec-push-back. *See also* [Feedback loop](#gloss-feedback-loop).

[**Prediction interval**]{#gloss-prediction-interval}
:   A range expected to contain a stated share of future observations. It is wider than a confidence interval, because it includes the noise in individual values as well as uncertainty in the estimate. Explained in @sec-learning. *See also* [Confidence interval](#gloss-confidence-interval).

[**Preset**]{#gloss-preset}
:   A named set of values for a molecule's constants, such as 'Oscillation (ignores it)', run with `run(preset=...)`. Explained in @sec-package. *See also* [Constant](#gloss-constant).

## R

[**R squared**]{#gloss-r-squared}
:   The share of the variation in the data that a fit accounts for. High values are easy to get on trending data and prove little on their own. Explained in @sec-learning. *See also* [Residuals](#gloss-residuals).

[**Reasoning model**]{#gloss-reasoning-model}
:   A language model that works through intermediate steps before it answers. Useful for critique, such as checking a polarity or explaining a test failure; its argument is then tested by simulation. Explained in @sec-ai-partners. *See also* [Large language model](#gloss-large-language-model), [Agent](#gloss-agent).

[**Red team**]{#gloss-red-team}
:   A deliberate attack on a chosen policy: scenarios within plausible ranges in which it might fail. In the book a reasoning model proposes them and code runs and screens each one. Explained in @sec-ai-decide. *See also* [Automation bias](#gloss-automation-bias).

[**Reference mode**]{#gloss-reference-mode}
:   The behaviour over time that a model sets out to explain, drawn from data or experience, such as a list that rises for a decade. Explained in @sec-story-to-diagram. *See also* [Dynamic hypothesis](#gloss-dynamic-hypothesis).

[**Regret**]{#gloss-regret}
:   How much worse a policy does in a scenario than the best policy for that scenario. A minimax-regret choice keeps the largest regret as small as possible. Explained in @sec-model-to-decision. *See also* [Robust policy](#gloss-robust-policy), [Scenario](#gloss-scenario).

[**Reinforcing loop**]{#gloss-reinforcing-loop}
:   A feedback loop that amplifies change, producing growth or collapse. It has an even number of negative links and is labelled R. Explained in @sec-push-back. *See also* [Balancing loop](#gloss-balancing-loop), [Exponential growth](#gloss-exponential-growth).

[**Replay file**]{#gloss-replay-file}
:   A file that keeps every request a chapter sends to a language model and the reply it got, keyed by a hash of the whole request, so the chapter runs offline and gives the same text every time. Explained in @sec-ai-partners. *See also* [Large language model](#gloss-large-language-model).

[**Residence time**]{#gloss-residence-time}
:   The average time a unit spends in a stock. In a steady state the stock equals its inflow times its residence time. Explained in @sec-loops. *See also* [Little's law](#gloss-little-s-law).

[**Residuals**]{#gloss-residuals}
:   The differences between data and model output. They should look like noise; a trend, a pattern or long runs on one side of zero mean the model is missing something. Explained in @sec-learning. *See also* [Autocorrelation](#gloss-autocorrelation).

[**Rework cycle**]{#gloss-rework-cycle}
:   A project structure in which some work is done wrong, the flaws stay hidden for a time, and the work returns to be done again when they are found. Explained in @sec-patterns.

[**Robust policy**]{#gloss-robust-policy}
:   A policy that performs acceptably across the plausible scenarios, not just in the most likely one. Explained in @sec-model-to-decision. *See also* [Scenario](#gloss-scenario), [Regret](#gloss-regret).

[**Root mean square error**]{#gloss-root-mean-square-error}
:   The square root of the average squared miss, in the data's own units (RMSE). It weights large misses more than the mean absolute error does. Explained in @sec-learning. *See also* [Mean absolute error](#gloss-mean-absolute-error).

## S

[**Scenario**]{#gloss-scenario}
:   One combination of assumptions about what the decision maker cannot control, used to test policies against several possible futures. Explained in @sec-model-to-decision. *See also* [Robust policy](#gloss-robust-policy).

[**Schedule pressure**]{#gloss-schedule-pressure}
:   The gap between the time needed to finish the work and the time left, which pushes people to work faster or longer, often at a cost in quality. Explained in @sec-decisions. *See also* [Effect](#gloss-effect).

[**Sensitivity analysis**]{#gloss-sensitivity-analysis}
:   Testing how much a model's results, its behaviour or its policy conclusions change when uncertain assumptions change. Explained in @sec-testing. *See also* [Monte Carlo analysis](#gloss-monte-carlo-analysis).

[**Shifting dominance**]{#gloss-shifting-dominance}
:   A change in which loop controls a stock over time, for example from reinforcing growth to balancing saturation in an S-curve. Explained in @sec-loops. *See also* [Loop dominance](#gloss-loop-dominance).

[**Shifting the burden**]{#gloss-shifting-the-burden}
:   An archetype in which a symptomatic fix is used instead of a fundamental solution and weakens the capacity to apply it. Explained in @sec-patterns. *See also* [Fixes that fail](#gloss-fixes-that-fail).

[**Smoothing**]{#gloss-smoothing}
:   Forming a belief by moving part of the way towards new information each period, so that it follows the information with a lag. Also called adaptive expectations or exponential smoothing. Explained in @sec-delays. *See also* [Information delay](#gloss-information-delay).

[**Spec**]{#gloss-spec}
:   The dictionary that describes a `loopbricks` model: its stocks, flows and variables with their equations. `lb.Model.from_spec` turns it into a model. Explained in @sec-package.

[**Standard deviation**]{#gloss-standard-deviation}
:   A measure of spread: roughly how far a typical value sits from the mean, in the data's own units. It is the square root of the variance. Explained in @sec-learning. *See also* [Normal distribution](#gloss-normal-distribution).

[**Standard error**]{#gloss-standard-error}
:   The standard deviation of an estimate across repeated samples; for a mean, the standard deviation of the data divided by the square root of the number of observations. Explained in @sec-learning. *See also* [Confidence interval](#gloss-confidence-interval).

[**Stock**]{#gloss-stock}
:   A quantity that accumulates, such as patients waiting or money in an account. It changes only through its flows, and it is the memory of the system. Drawn as a box. Explained in @sec-push-back. *See also* [Flow](#gloss-flow), [Net flow](#gloss-net-flow).

[**Stock management**]{#gloss-stock-management}
:   The task of keeping a stock at a target by adjusting an inflow that arrives after a delay; the decision rule replaces expected outflow and corrects the gaps in the stock and the supply line. Explained in @sec-swings. *See also* [Supply line](#gloss-supply-line), [Supply line weight](#gloss-supply-line-weight).

[**Stock-and-flow diagram**]{#gloss-stock-and-flow-diagram}
:   A diagram of stocks (boxes), flows (pipes with valves), sources and sinks outside the model (clouds), and the information links that set each flow. Explained in @sec-push-back. *See also* [Causal loop diagram](#gloss-causal-loop-diagram).

[**Success to the successful**]{#gloss-success-to-the-successful}
:   An archetype in which two parties compete for a shared resource and the one that gets ahead gets more of it, so the gap widens. Explained in @sec-patterns. *See also* [Path dependence](#gloss-path-dependence).

[**Supply line**]{#gloss-supply-line}
:   The stock of items ordered but not yet received. Decision makers who ignore it keep ordering while goods are already on the way. Explained in @sec-swings. *See also* [Acquisition delay](#gloss-acquisition-delay), [Stock management](#gloss-stock-management).

[**Supply line weight**]{#gloss-supply-line-weight}
:   How much of the supply line gap an ordering rule takes into account, from 0 (ignored) to 1 (fully counted). Low weights make a stock overshoot and oscillate. Explained in @sec-swings. *See also* [Supply line](#gloss-supply-line).

[**Surrogate model**]{#gloss-surrogate-model}
:   A fast approximation of a slow model, fitted to a sample of its runs and used to explore many settings; also called a metamodel or emulator. Its answers must be confirmed with the real model. Explained in @sec-ai-decide. *See also* [Machine learning](#gloss-machine-learning).

[**Sycophancy**]{#gloss-sycophancy}
:   A language model's tendency to agree with the framing of a question, so that a leading question gets the answer it leads to. Ask neutral questions and let a simulation decide. Explained in @sec-ai-partners. *See also* [Hallucination](#gloss-hallucination).

[**Synthetic data**]{#gloss-synthetic-data}
:   Data generated from a model with known settings and added noise. Because the truth is known, it tests whether an estimation method can recover it. Explained in @sec-calibration. *See also* [Calibration](#gloss-calibration).

[**Synthetic persona**]{#gloss-synthetic-persona}
:   A language model playing a stakeholder, used to rehearse an interview. What a persona says is rehearsal, not evidence, and never enters the edge list. Explained in @sec-ai-conceptualise. *See also* [Causal link](#gloss-causal-link).

[**System dynamics**]{#gloss-system-dynamics}
:   A method for understanding how the structure of a system, its stocks, flows, feedback loops and delays, produces its behaviour over time, and for testing policies in a simulation model. Explained in @sec-push-back.

## T

[**Time constant**]{#gloss-time-constant}
:   The time a stock would take to empty if its outflow stayed at its first value. After one time constant an exponential decay has covered about 63% of the way, after three about 95%. Explained in @sec-loops. *See also* [Exponential decay](#gloss-exponential-decay), [Adjustment time](#gloss-adjustment-time).

[**Time horizon**]{#gloss-time-horizon}
:   The span of time a model must cover to show the problem and the effects of policies on it, long enough for the slowest important delays to play out. Explained in @sec-story-to-diagram.

[**Tipping point**]{#gloss-tipping-point}
:   A threshold beyond which a reinforcing loop takes over and carries the system to a different state. Explained in @sec-loops. *See also* [Path dependence](#gloss-path-dependence), [Equilibrium](#gloss-equilibrium).

[**Tool**]{#gloss-tool}
:   A Python function an agent may call, described to the model by its signature and docstring. It does the computing; the model only chooses which tool to call and with what. Explained in @sec-ai-partners. *See also* [Agent](#gloss-agent).

[**Tragedy of the commons**]{#gloss-tragedy-of-the-commons}
:   An archetype in which many users each gain by taking more of a shared resource, until together they deplete it. Explained in @sec-patterns. *See also* [Archetype](#gloss-archetype).

[**Training and test sets**]{#gloss-training-and-test-sets}
:   The part of the data a model is fitted on, and the part held back to measure its error. For time series, train on the past and test on what came after. Explained in @sec-learning. *See also* [Out-of-sample test](#gloss-out-of-sample-test), [Leakage](#gloss-leakage).

## U

[**Unsourced number**]{#gloss-unsourced-number}
:   A number in an AI answer or brief that no tool result, run or input contains. The book's checks flag every one. Explained in @sec-ai-partners. *See also* [Hallucination](#gloss-hallucination), [Agent](#gloss-agent).

## V

[**Validation**]{#gloss-validation}
:   Building confidence that a model is fit for its purpose, through tests of its structure and of its behaviour. No model is ever proved valid. Explained in @sec-testing. *See also* [Verification](#gloss-verification).

[**Verification**]{#gloss-verification}
:   Checking that the code computes what the equations say: units, conservation, time step and the arithmetic. Explained in @sec-testing. *See also* [Validation](#gloss-validation).
